```python
import xarray as xr

# Periods
ds1 = xr.open_dataset("Peak_Period_old.nc")
ds2 = xr.open_dataset("Peak_2015.nc")

ds_swell = xr.concat([ds1, ds2], dim="valid_time")
ds_swell.to_netcdf("Peak_Period.nc")
```

```python
from netCDF4 import Dataset
nc1 = Dataset('Swell_waves.nc', 'r')
nc2 = Dataset('Wind_waves.nc', 'r')
nc3 = Dataset('Combined_waves.nc', 'r')
nc4 = Dataset('Peak_Period.nc', 'r')
pass

lon1 = nc1.variables['longitude']
lat1 = nc1.variables['latitude']
time1 = nc1.variables['valid_time']

lon2 = nc2.variables['longitude']
lat2 = nc2.variables['latitude']
time2 = nc2.variables['valid_time']

lon3 = nc3.variables['longitude']
lat3 = nc3.variables['latitude']
time3 = nc3.variables['valid_time']

lon4 = nc4.variables['longitude']
lat4 = nc4.variables['latitude']
time4 = nc4.variables['valid_time']

print('test nc1')

for d in nc1.dimensions.items():
    print(d)

print('test nc2')

for d in nc2.dimensions.items():
    print(d)

print('test nc3')

for d in nc3.dimensions.items():
    print(d)

print('test nc4')

for d in nc4.dimensions.items():
    print(d)
```

```python
lon1_array = lon1[:]
lat1_array = lat1[:]
time1_array = time1[:]

lon2_array = lon2[:]
lat2_array = lat2[:]
time2_array = time2[:]

lon3_array = lon3[:]
lat3_array = lat3[:]
time3_array = time3[:]

lon4_array = lon4[:]
lat4_array = lat4[:]
time4_array = time4[:]

print(lon1_array)
print(lat1_array)
print(time1_array)

print(lon2_array)
print(lat2_array)
print(time2_array)

print(lon3_array)
print(lat3_array)
print(time3_array)

print(lon4_array)
print(lat4_array)
print(time4_array)
pass
```

```python
hsswell = nc1.variables['shts']
hswind = nc2.variables['shww']
hscombined = nc3.variables['swh']
peakperiod = nc4.variables['pp1d']

hsswell_array = hsswell[:]
hswind_array = hswind[:]
hscombined_array = hscombined[:]
peakperiod_array = peakperiod[:]

print("Hs swell array")
print(hsswell_array)

print("Hs wind array")
print(hswind_array)

print("Hs combined array")
print(hscombined_array)

print("Peak period array")
print(peakperiod_array)
```

```python
i = 1 #for the present case, the values can be 0,1
j = 1 #for the present case, the values can be 0,1
hswind_array = hswind[:,i,j]
hsswell_array = hsswell[:,i,j]
hscombined_array = hscombined[:,i,j]
peakperiod_array = peakperiod[:,i,j]

print("Hs wind array")
print(hswind_array)

print("Hs swell array")
print(hsswell_array)

print("Hs combined array")
print(hscombined_array)

print("Peak period array")
print(peakperiod_array)
```

```python
import pandas as pd
import numpy as np

data1 = np.array([time1_array,
         hsswell_array])
data1 = data1.T

df1 = pd.DataFrame(data1, columns=['time', 'Hs_swell'])

df1.to_csv('hs1arrays.csv', index=False)

data2 = np.array([time2_array,
         hswind_array])
data2 = data2.T

df2 = pd.DataFrame(data2, columns=['time', 'Hs_wind'])

df2.to_csv('hs2arrays.csv', index=False)

data3 = np.array([time3_array,
         hscombined_array])
data3 = data3.T

df3 = pd.DataFrame(data3, columns=['time', 'Hs_combined'])

df3.to_csv('hs3arrays.csv', index=False)

data4 = np.array([time4_array,
         peakperiod_array])
data4 = data4.T

df4 = pd.DataFrame(data4, columns=['time', 'Peak_period'])

df4.to_csv('peakperiodarrays.csv', index=False)
```

```python
import pandas as pd

df1 = pd.read_csv("hs1arrays.csv")
df2 = pd.read_csv("hs2arrays.csv")
df3 = pd.read_csv("hs3arrays.csv")
df4 = pd.read_csv("peakperiodarrays.csv")

df1["time"] = pd.to_datetime(df1["time"], unit="s")
df2["time"] = pd.to_datetime(df2["time"], unit="s")
df3["time"] = pd.to_datetime(df3["time"], unit="s")
df4["time"] = pd.to_datetime(df4["time"], unit="s")

df_total = df1.merge(df2, on="time", how="inner")
df_total = df_total.merge(df3, on="time", how="inner")
df_total = df_total.merge(df4, on="time", how="inner")

df_total.to_csv("total_data.csv", index=False)

print(df_total.head())
```
